```haskell
primerElemento :: [a] -> a
primerElemento x = head x
primerElemento [5,8,9,6,3]
```

```haskell
productoLista :: [Int] -> Int
productoLista[] = 1
productoLista(x:xs)= x * productoLista xs
productoLista[2,4,5,6]
```

```haskell
eliminarNegativos :: [Int] -> [Int]
eliminarNegativos [] = []
eliminarNegativos (x:xs)
       | x<0 = eliminarNegativos xs
       | otherwise = x : eliminarNegativos xs
eliminarNegativos [1,5,-6,8,9,7,-9]
```

```haskell
invertir :: [a] -> [a]
invertir [] = []
invertir (x:xs) = invertir xs ++ [x]
invertir [1,2,3,4]
```

```haskell
sumaCuadrados :: Float -> Float -> Float
sumaCuadrados x y = (x**2)+(y**2)
sumaCuadrados 5 2
```

```haskell
dobleLista :: [Int] -> [Int]
dobleLista xs = map(*2) xs
dobleLista[2,3,4,5]
```

```haskell
dobleLista :: [Int] -> [Int]
dobleLista [] = []
dobleLista (x:xs) = (x*2) : dobleLista xs
dobleLista[2,3,4,5]
```

```haskell
clasificarNota :: Int -> String
clasificarNota x
   | x >= 4 = "Aprobado"
   | x >= 3 = "En recuperación"
   | otherwise = "Reprobado"
clasificarNota 5
```

```haskell
fibonacci :: Int -> Int
fibonacci 0 = 0
fibonacci 1 = 1
fibonacci n = fibonacci (n-1) + fibonacci (n-2)
fibonacci 6
```
